```python
from insightface.app import FaceAnalysis
import cv2

app = FaceAnalysis(name='buffalo_l')  # ArcFace pretrained pack
app.prepare(ctx_id=-1)  # 0 = GPU, -1 = CPU
```

```python
import shutil
```

```python
import os
import subprocess
import cv2
import pandas as pd
from pathlib import Path

remote_base = "rclone2.0:Timesformer"
output_base = Path("Ravdess_Spatial_Embeddings")
output_base.mkdir(exist_ok=True)

processed_videos = set(processed_videos)

for video in videos:
    video = video.rstrip("/")

    if video in processed_videos:
        print(f"Embeddings already extracted for {video}")
        continue

    print(f"Processing video: {video}")

    remote_video_path = f"{remote_base}/{video}"

    local_video_path = Path(f"/tmp/{video}")
    local_video_path.mkdir(parents=True, exist_ok=True)

    # 1. COPY VIDEO FROM DRIVE
    result = subprocess.run([
        "rclone", "copy",
        remote_video_path,
        str(local_video_path)
    ])

    if result.returncode != 0:
        print(f"Failed to copy {video}")
        continue

    # 2. SORT FRAMES (VERY IMPORTANT)
    all_frames = sorted(local_video_path.glob("*.png"))

    embeddings = []

    for frame_path in all_frames:
        print(f"Processing frame: {frame_path.name}")

        img = cv2.imread(str(frame_path))

        if img is None:
            continue

        faces = app.get(img)

        print(f"Number of faces detected: {len(faces)}")

        for face in faces:
            embeddings.append(face.embedding)

    # 3. SAVE OUTPUT
    df = pd.DataFrame(embeddings)
    # video_store_path = output_base / video
    # df.to_csv(output_base + video + ".csv", index=False)
    os.makedirs("Ravdess_Spatial_Embeddings", exist_ok=True)

    csv_path = os.path.join(
        "Ravdess_Spatial_Embeddings",
        f"{video}.csv"
    )
    
    df.to_csv(csv_path, index=False)

    # 4. CLEANUP
    shutil.rmtree(local_video_path, ignore_errors=True)
```
