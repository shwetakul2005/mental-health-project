```python
import os
from pathlib import Path

BASE_DIR = Path("/content/drive/MyDrive")

# Change this after checking your Drive structure
PROCESSED_DIR = BASE_DIR / "DAIC_WOZ_VIDEO_BRANCH" / "data" / "processed"

print("Exists:", PROCESSED_DIR.exists())
print("Path:", PROCESSED_DIR)
```

```python
import pandas as pd
```

```python
video_features_df = pd.read_csv(PROCESSED_DIR / "video_session_features.csv")
labels_clean = pd.read_csv(BASE_DIR / "DAIC_WOZ_VIDEO_BRANCH" / "data" / "labels" / "labels_train_dev_clean.csv")

video_dataset = video_features_df.merge(labels_clean, on="Participant_ID", how="inner")

print(video_dataset.shape)
video_dataset.head()
```

```python
print("Video participants:", video_features_df["Participant_ID"].nunique())
print("Label participants:", labels_clean["Participant_ID"].nunique())
print("Matched participants:", video_dataset["Participant_ID"].nunique())

print(video_dataset["split"].value_counts())
print(video_dataset["PHQ_Binary"].value_counts())
```

```python
video_dataset.to_csv(PROCESSED_DIR / "video_dataset_with_labels.csv", index=False)
```
